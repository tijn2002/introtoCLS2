```python
import  numpy               as np
import  matplotlib.pyplot   as plt
from    GA                  import *
```

```python
base = {"beta": 0.5, "gamma": 0.1, "mu": 0.01, "N0": 1000}
state0 = [999, 1, 0]
t_max = 150
n_runs = 100
t_grid = np.linspace(0, t_max, 500)
names = ["S", "I", "R"]
```

### Changing parameter $\beta$

```python
name = "beta"
values = [0.2, 0.3, 0.5, 0.8]
colors = plt.cm.viridis(np.linspace(0, 0.9, len(values)))

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for value, color in zip(values, colors):
    params = {**base, name: value}
    runs = run_ensemble(state0, params, t_max, n_runs, t_grid)
    _, states_ode = deterministic(state0, params, t_max, len(t_grid))

    for i in range(3):
        axes[i].plot(t_grid, runs[:, :, i].mean(axis=0), color=color,
                     label=f"{name}={value}")
        axes[i].plot(t_grid, states_ode[:, i], color=color, linestyle="--",
                     linewidth=1)

for ax, compartment in zip(axes, names):
    ax.set_xlabel("time")
    ax.set_ylabel(compartment)
    ax.set_title(compartment)

axes[2].legend(fontsize=7, loc="center left", bbox_to_anchor=(1.02, 0.5))
plt.tight_layout()
plt.show()
```

```python
betas = [0.2, 0.3, 0.5, 0.8]
colors = plt.cm.viridis(np.linspace(0, 0.9, len(betas)))

fig, ax = plt.subplots(figsize=(6, 4))

for beta, color in zip(betas, colors):
    params = {**base, "beta": beta}
    runs = run_ensemble(state0, params, t_max, n_runs, t_grid)
    cov = covariance(runs, 0, 1)
    ax.plot(t_grid, cov, color=color, label=f"beta={beta}")

ax.axhline(0, color="k", linewidth=0.5)
ax.set_xlabel("time")
ax.set_ylabel("Cov(S, I)")
ax.set_title("Cov(S, I)")
ax.legend(fontsize=7)
plt.tight_layout()
plt.show()
```
